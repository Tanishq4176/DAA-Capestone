## Observations and future optimization

- Sorting stays cheap (O(n log n) Timsort).
- Knapsack grows with both parcel count and capacity (O(nC)).
- Routing dominates because nearest-neighbor repeats Dijkstra for each stop.

Possible improvements: 1D DP when reconstruction is handled separately, a cached k×k distance matrix on selected stops, two-opt on the tour, and more benchmark repeats across machines.
